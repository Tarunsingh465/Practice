# Inheritance in python
### Inheritance is a fundamental concept in Object-Oriented Programming(OOP) that allows a class to inherit attributes and methods from another class. this lesson covers single ineritence and multiple inheritance, demonstrating how to create and use them in Python.

### First let's code and understand the single Inheritance 

```python
class Car:
    def __init__(self,windows,doors,enginetype):
        self.windows=windows
        self.doors=doors
        self.enginetype=enginetype

    def drive(self):
        return (f"The person will drive the {self.enginetype} car.")      
```

```python
car1=Car(6,4,'Petrol')
print(car1.drive())
```

```python
class Tesla(Car):
    def __init__(self,windows,doors,enginetype,is_selfdriving):
        super().__init__(windows,doors,enginetype)
        self.is_selfdriving=is_selfdriving
    
    def selfdriving(self):
        return f'Tesla supports self driving : {self.is_selfdriving}'
    
tesla=Tesla(6,4,"Electric",True)
print(tesla.selfdriving())
print(tesla.drive())
```

```python
class Person:
    def __init__(self,name,age,city):
        self.name=name
        self.age=age
        self.city=city
    
    def marital_status(self):
        return f"{self.name} is not married."
```

```python
p1=Person("Tarun",21,'Indore')
print(p1.marital_status())
```

```python
class Student(Person):
    def __init__(self, name, age, city,qualification,employment_status):
        super().__init__(name, age, city)
        self.qualification=qualification
        self.employment_status=employment_status
    def report(self):
        return f"Name of the student is {self.name}. He is a sincere student."
    
    def status(self):
        return f'{self.name} has completed the {self.qualification} and got a position in a company as {self.employment_status}. '

s1=Student('Tarun',21,'Indore', 'Data Science Engineering', 'Intern')
print(s1.report())
print(s1.status())
print(s1.marital_status())
```

```python
class Dog:
    def __init__(self,name):
        self.name=name
    
    def bark(self):
        return f"{self.name} says woof"
    
d1=Dog('Tommy')
print(d1.name)
print(d1.bark())
```

```python
class Husky(Dog):
    def __init__(self, name,breed,color):
        super().__init__(name)
        self.breed=breed
        self.color=color

    def fav_play(self):
        return f"{self.breed} loves winter and the color of my dog is {self.color}."

h1=Husky('Snowy','Husky','Grey')
print(h1.bark())
print(h1.fav_play())
```

### Let's jump onto the multiple inheritance and understand it wrt to its code

#### In this a class inherits the property of more than one class

```python
# Base class - 1
class Animal:
    def __init__(self,name):
        self.name=name
    
    def null(self):
        return f"Subclass must implement this method"


# Base class - 2 
class Pet:
    def __init__(self,owner):
        self.owner=owner
    

# Derived Class
class Dog(Animal,Pet):
    def __init__(self, name, owner):
        Animal.__init__(self, name)
        Pet.__init__(self, owner)
    
    def speak(self):
        return f"{self.name} says woof"
```

```python
# Creating am object

dog=Dog('Ragnar','Tarun')
print(dog.null())
print(dog.speak())
```
