```python
class Dog:
    ## constructor
    def __init__(self,name,age):
        self.name=name
        self.age=age
dog1=Dog('Buddy',3)
print(dog1)
print(dog1.name)
print(dog1.age)
```

```python
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    
    def greet(self):
        print(f"Hello, my name is {self.name}")

p1=Person('Tarun',21)
print(p1.name)
p1.greet()
```

```python
class Car:
    def __init__(self,brand,model):
        self.brand=brand
        self.model=model
    def mycar(self):
        print(f'My car brand is {self.brand} and the model is {self.model}.')

c1=Car('Porche',911)
print(c1.brand)
print(c1.model)
c1.mycar()
```

```python
class Person:
    def __init__(self,name,age,city):
        self.name=name
        self.age=age
        self.city=city
        
p1=Person('Tarun',21,"Indore")

print(f'My name is {p1.name}, my age is {p1.age} and i live in {p1.city}.')
```

```python
class Person:
    def __init__(self,name,age=18):
        self.name=name
        self.age=age
p1=Person("Raj")
p2=Person('Tarun',21)
print(p1.name, p1.age)
print(p2.name, p2.age)
```

```python
class Person:
    def __init__(self,name,age,city,country):
        self.name=name
        self.age=age
        self.city=city
        self.country=country

p1=Person('Tarun',21,'Indore','India')
print(p1.name)
print(p1.age)
print(p1.city)
print(p1.country)
print(f"Hey guys! My name is {p1.name}, i am {p1.age} years old and i live in {p1.city}, {p1.country}.")
```

```python
class Person:
    def __init__(self,name):
        self.name=name
        
    def printname(self):
        print(self.name)
        
p1=Person("Ram")
p2=Person("Shyam")

p1.printname()
p2.printname()
```
