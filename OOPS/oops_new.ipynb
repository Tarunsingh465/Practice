```python
class BankAccount:
    def __init__(self,owner,balance=0):
        self.owner=owner
        self.balance=balance
    
    def deposit(self,amount):
        self.balance+=amount
        print(f'Amount {amount} is deposited and your new balance is {self.balance}')
    
    def withdraw(self,amount):
        if self.balance<amount:
            print('Insufficient Amount')
        
        else:
            self.balance-=amount
            print(f'Amount {amount} is withdrawn and your new balance is {self.balance}')
            
    def account_balance(self):
        return self.balance
    
a=BankAccount('Tarun',100000)

```

```python
a.deposit(5000)    
```

```python
a.withdraw(3000)
```

```python
a.account_balance()
```

```python
class Account:
    def __init__(self,holder,balance):
        self.holder=holder
        self.balance=balance
    
    def deposit(self,amount):
        self.balance+=amount
        return (f"${amount} is deposited by {self.holder}. New balance in the account is ${self.balance}")
    
    def withdraw(self,amount):
        if amount>self.balance:
            return (f"Insufficient Amount in {self.holder}'s account!!!")
        else:
            self.balance-=amount
            return (f"${amount} is withdrawn by {self.holder}. New balance is ${self.balance}.")
    
    def account_balance(self):
        return f"The available account balance in {self.holder}'s account is ${self.balance}"
    
a1=Account('Tarun',5000)
print(a1.account_balance())
print(a1.deposit(500))
print(a1.withdraw(300))
print(a1.account_balance())
```
