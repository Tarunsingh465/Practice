```python
class Calculator:
    def add(self,a,b):
        return a+b
    
    def multiply(self,a,b):
        return a*b
    
    def subtraction(self,a,b):
        return b-a
    
    def division(self,a,b):
        return b/a

calc=Calculator()
print(calc.add(5,3))
print(calc.multiply(5,3))
print(calc.subtraction(5,3))
print(calc.division(5,3))
```

```python
class Calculator:
    def __init__(self,a,b):
        self.a=a
        self.b=b
    def add(self):
        return self.a+self.b
    
    def multiply(self):
        return self.a*self.b
    
    def subtraction(self):
        return self.b-self.a
    
    def division(self):
        return self.b/self.a

calc=Calculator(5,3)
print(calc.add())
print(calc.multiply())
print(calc.subtraction())
print(calc.division())
```
