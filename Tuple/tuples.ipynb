### Immutable nature of tuples

#### Tuples are Immutable, it means the values can't be changed once assigned

```python
lst=[1,2,3,4,5]
print(lst)

lst[1]='Tarun'
print(lst)
```

```python
''' When we try to do the above operation with tuple it behaves differently because of its 
immutable nature.'''

numbers[2]='Tarun'

```

### Tuple Methods

```python
# Count function helps us to count the occurence of any specific element

print(numbers.count(1))

# Index function helps us to find the element on that particular index
print(numbers.index(3))
```

```python
# Packing nd unpacking tuple

packed_tuple=1,"Hello",3.14
print(packed_tuple)
```

```python
# unpacking a tuple

a,b,c=packed_tuple
print(a)
print(b)
print(c)
```

```python
# unpacking with a *

numbers=(1,2,3,4,5,6)
first,*middle,last=numbers
print(first)
print(middle)
print(last)
```

```python
# Nested tuple

lst=[[1,2,3,4],[5,6,7,8],[4,"Tarun"]]
lst[0]
lst[0][3]
lst[1][:2]
```

```python
lst=[[1,2,3,4],[5,6,7,8],(4,"Tarun",True)]
lst[2][::-1]
```
