```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n*factorial(n-1)
factorial(5)
```

```python
def countupto(n):
    count=1
    while count <= n:
        yield count
        count+=1
for num in countupto(10):
    print(num)
```
