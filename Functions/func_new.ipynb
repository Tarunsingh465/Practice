```python
def changecase(func):
    def myinner():
        return func().upper()
    return myinner

@changecase
def myfunction():
    return 'Hello Sally'

print(myfunction())
```

```python
def greet():
    return 'Hello'

def decorator(func):
    def wrapper():
        print("Good morning!")
        return func()
    return wrapper

greet = decorator(greet)

result = greet()

print(result)
```

```python
def changecase(func):
    def myinner():
        return func().upper()
    return myinner

@changecase
def name():
    return ('Tarun')

@changecase
def surname():
    return ('Singh')

print(name())
print(surname())
```

```python
def talk(func):
    def wrapper(x):
        print('Nice to meet you!')
        return func(x)
    return wrapper

@talk
def greet(name):
    return f'Hello, {name}'

print(greet("Tarun"))
```

```python
def changecase(func):
    def myinner(x):
        print(func(x).upper())
        return ("Nice to meet you")
    return myinner

@changecase
def myfunction(nam):
    return "Hello! " + nam

print(myfunction("tarun"))
```

```python
def changecase(func):
    def myinner(*args,**kwargs):
        return func(*args,**kwargs).upper()
    return myinner

@changecase
def myfunction(nam,surname='Singh'):
    return f"Hello, {nam} {surname}"
print(myfunction('Tarun'))
```

```python
def changecase(n):
    def changecase(func):
        def myinner():
            if n==1:
                a=func().lower()
            else:
                a=func().upper()
            return a
        return myinner
    return changecase

@changecase(2)
def myfunction():
    return "Hello Tarun"

print(myfunction())
```

```python
import functools
def changecase(func):
    @functools.wraps(func)
    def myinner():
        return func().upper()
    return myinner

@changecase
def myfunction():
    return "Have a great day!"
print(myfunction.__name__)
```
