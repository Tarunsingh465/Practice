```python
import math

# returns the max or min value of the given numbers
x=min(12,45,34)
y=max(12,45,34)
print(f'Minimum number: {x}\nMaximum number: {y}')
```

```python
# converts the numbre into absolute positive number

x = abs(-7.25)

print(x)
```

```python
# returns the calculated value of a number after applying the power

x = pow(2, 3)

print(x)
```

```python
# returns the square root of the given number

x=(math.sqrt(49))
print(x)
```

```python
''' The math.ceil() method rounds a number upwards to its nearest integer and
the math.floor() method rounds a number downwards to its nearest integer'''

x = math.ceil(1.4)
y = math.floor(1.4)

print(x) # returns 2
print(y) # returns 1
```

```python
# returns the value of pi

x = math.pi

print(x)
```
