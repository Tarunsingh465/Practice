## \d → digit
## \w → word character
## \s → whitespace

```python
import re

text = "I am learning Python"

result = re.search("Python", text)

print(result)
```

```python
result = re.search("dog", "I have a cat")

print(result)
```

```python
import re

text = "My age is 21"

result = re.search(r"\d", text)

print(result)
```

```python
import re

text = "My age is 21"

result = re.search(r"\d+", text)

print(result.group())
```

```python
text = "Hello 123"
```

```python
# \d
re.findall(r'\d',text)
```

```python
# \w
re.findall(r"\w", text)
```

```python
# \s
re.findall(r"\s", text)
```

```python
# search() finds the first match.
result=re.search(r"\d+", "I have 21 apples and 15 oranges")
print(result.group())
```

```python
# use findall()

re.findall(r"\d+", "I have 21 apples and 15 oranges")
```
