```python
# Assignment 3

set1=set(range(1,6))
set2=set(range(2,11,2))

# Union

print(f'Union set:{set1.union(set2)}')

# Intersection

print(f'Intersection set:{set1.intersection(set2)}')

# Difference

print(f'Difference set:{set1.difference(set2)}')

# Symmetric difference

print(f'Symmetric difference of both the sets:{set1.symmetric_difference(set2)}')
```
