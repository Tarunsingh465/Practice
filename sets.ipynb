```python
# creating a set

my_set={1,2,3,4,5}
print(my_set)
print(type(my_set))
```

```python
my_empty_set=set()
print(type(my_empty_set))
my_empty_set
```

```python
# typecasting with the help of set

my_set=set([1,2,3,4,5,6])
print(my_set)
```

```python
# No duplicates allowed

my_set=set([1,2,3,4,5,6,6,3,4,5])
print(my_set)
```

## Basic Set operations

```python
# Adding and removing elements

my_set
```

```python
my_set.add(7)
my_set
```

```python
# removing elements

my_set.remove(3)
print(my_set)
```

```python
# If an element is not present in a set and we try to remove it then it throws an error

my_set.remove(10)
my_set
```

```python
# But if instead of remove we use discard func then it can be resolved

my_set.discard(10)
print(my_set)
```

```python
# Removes the first element which was inserted in the set
# FIFO- First In First Out

removed_element=my_set.pop()
print(removed_element)
print(my_set)
```

```python
# Membership test

print(4 in my_set)
print(9 in my_set)
```

```python
# Mathematical operation in sets

set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}

# Union

union_set=set1.union(set2)
print(union_set)

# Intersection

intersection_set=set1.intersection(set2)
print(intersection_set)
```

```python
set1.intersection_update(set2)
set1
```

```python
set1
```

```python
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}

# Set difference

print(set1.difference(set2))
```

```python
# Symmetric difference removes only common elements

set1.symmetric_difference(set2)
```

```python
# Sets methods

set1={1,2,3}
set2={3,4,5}

# issubset method

print(set1.issubset(set2))

# issuperset

print(set1.issuperset(set2))
```

```python
text="Hey guys my name is Tarun"
words=text.split()

# converts list of words to set to get unique words

unique_words=set(words)
print(unique_words)
print(len(unique_words))
```
