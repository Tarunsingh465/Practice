# Organizing the Students' Grades 

```python
grades=[89,87,98,78,93]

# Adding a new grade
grades.append(85)

# Calculating the average grade
avg_grade= sum(grades)/len(grades)
print(f'Average Grade: {avg_grade:.2f}')

# Finding the highest and lowest grade
highest_grade=max(grades)
lowest_grade=min(grades)
print(f'Highest grade: {highest_grade}')
print(f'Lowest grade: {lowest_grade}')
```

# Managing Inventories

```python
# Items in the inventory
inventory=['Apples','Oranges','Bananas','Grapes']

# Adding a new item
inventory.append("Strawberries")

# Removing an item that is out of stock
inventory.remove("Bananas")

# Checking if an item is in stock
if 'Bananas' in inventory:
    print("Bananas are in stock")
else:
    print("Bananas are not in stock")
    
# Printing the inventory
print("Items availiable in inventory are as follows:")
for item in inventory:
    print(f"-{item}")
```
