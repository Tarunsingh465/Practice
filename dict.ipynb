```python
# shallow copy

student_copy=student
print(student_copy)
print(student)
```

```python
# Chnaging some values
student['name']='Sensei'
print(student)
print(student_copy)

# Here even if I change the value in the original var it will change it in copy var
```

```python
# To deal with this problem we need shallow copy

student_copy1=student.copy() ## Shalow copy
print(student_copy1)
print(student)
```

```python
student["name"]="Raj"
print(student)
print(student_copy) # It changed the value wrt to student
print(student_copy1) # It remain the same as it was created
```

```python
# Nested Dictionaries
students={
    "student1":{"name":"Tarun","age":21},
    "student2":{"name":"Sensei","age":18},
    "student3":{"name":"Raj","age":20}
}
print(students)
```

```python
# Access nested dictionaries elelment

print(students['student1']['name'])
print(students['student1']['age'])
```

```python
# Iterating over nested dictionaries

for student_id,student_info in students.items():
    print(f'{student_id}:{student_info}')
    for key,value in student_info.items():
        print(f'{key}:{value}')
```

### Dictionary comprehension

```python
# Dictionary comprehension

squares={x:x**2 for x in range(6)}
print(squares)
```

```python
# Conditional dictioanry comprehension

even={x:x**2 for x in range(11) if x%2==0}
print(even)
```

```python
# Use a dictioanry to count the frequency of elements in list

numbers=[1,2,2,3,3,3,4,4,4,4]
frequency={}

for number in numbers:
    if number in frequency:
        frequency[number]+=1
    else:
        frequency[number]=1
print(frequency)
```

```python
# Merge two dictionaries into single one

dict1={'a':1,'b':2}
dict2={'c':3,'d':4}

merged_dict={**dict1,**dict2}
print(merged_dict)
```
