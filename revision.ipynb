```python
def changecase(func):
    def wrapper():
        return func().upper()
    return wrapper

@changecase
def greet():
    return 'Hello Tarun'

x=greet()
print(x)
```

```python
def count(n):
    if n<=0:
        return 0
    else:
        count(n-1)
        print(n)
count(10)
```

```python
def count_rev(n):
    if n<=0:
        return 'Done'
    else:
        print(n)
        count_rev(n-1)

count_rev(10)
```
