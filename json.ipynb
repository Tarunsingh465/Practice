```python
import json

# json to python
# some JSON:
x= '''{"name": "John","age": 30,"city": "New York","isStudent": false,"hobbies": ["reading", "hiking"]}'''
y=json.loads(x)
print(y)
```

```python
# convert to json from any python object; in this case it is dictionary

x= {
    'name' : 'Tarun',
    'age' : 21,
    'city' : 'Indore',
    'country' : 'India'
}

# convert into json:
y=json.dumps(x)

# print the json result
print(y)
```
