```python
try:
    num=int(input("Enter a number: "))
    result=10/num

except ValueError:
    print("This is not a valid number")

except ZeroDivisionError:
    print("Enter denominator other than 0")
    
except Exception as e:
    print(e)
```

```python
try:
    num=int(input("Enter a number: "))
    result=100/num

except ValueError:
    print("This is not a valid number")

except ZeroDivisionError:
    print("Enter denominator other than 0")
    
except Exception as e:
    print(e)

else:
    print(f'The result is : {result:.2f}')
    
finally:
    print('Execution Completed!')
```
